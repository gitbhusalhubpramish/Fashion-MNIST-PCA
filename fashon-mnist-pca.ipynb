```python
rng = np.random.default_rng(42)
v = rng.random((785,1))
v = v.T@c
v = v/np.linalg.norm(v)
v
```

```python
v = np.array(v)
lam = np.array(v[0].T@c@v[0])
for i in range(len(c)-1):
    vi = v[i].T@c
    lam=np.append(lam,vi.T@c@vi)
    v=np.append(v,[vi/np.linalg.norm(vi)], axis=0)
```
